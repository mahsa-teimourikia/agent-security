# Multi-agent delegation security lab

This credential-free exercise proves authority attenuation before a child agent can use a tool.

```python
from pathlib import Path
import sys
from datetime import datetime, timedelta, timezone
sys.path.insert(0, str(Path.cwd().parents[1] / 'shared'))
from runtime_security_lab import Delegation, Run
now = datetime(2026, 8, 11, tzinfo=timezone.utc)
run = Run('team-run', 'north', 'state-v1', 'policy-v1')
```

## Authorized specialist

The research specialist receives only `read`, a strict subset of the parent’s scopes.

```python
safe = Delegation('orchestrator', 'researcher', 'north', frozenset({'read','write'}), frozenset({'read'}), now + timedelta(minutes=5))
assert run.delegate(safe, now=now)
```

## Privilege-escalation bypass

A child requesting `write` where the parent only has `read` is denied and traced.

```python
attack = Delegation('orchestrator', 'operator', 'north', frozenset({'read'}), frozenset({'read','write'}), now + timedelta(minutes=5))
assert not run.delegate(attack, now=now)
run.trace
```
