Use an offline runtime to validate that a compromised run cannot create new writes after revocation.

## Normal action then containment

The initial external action receives a receipt; containment changes the run to read-only before a new action can be attempted.

```python
assert run.commit_once('ticket-1')
run.revoke('writes')
assert run.status == 'read_only'
assert not run.commit_once('ticket-2')
run.trace
```
